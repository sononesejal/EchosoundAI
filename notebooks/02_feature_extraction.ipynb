```python
import librosa
import numpy as np
import pandas as pd
import os
from tqdm import tqdm

ESC_CSV = r"C:\Users\sonon\OneDrive\Documents\USHI_Project\dataset\ESC-50\esc50.csv"
AUDIO_PATH =r"C:\Users\sonon\OneDrive\Documents\USHI_Project\dataset\ESC-50\audio\audio"

def extract_features(file_path):
    audio, sr = librosa.load(file_path, duration=5)

    mfcc = np.mean(librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=13), axis=1)
    chroma = np.mean(librosa.feature.chroma_stft(y=audio, sr=sr), axis=1)
    rms = np.mean(librosa.feature.rms(y=audio))
    zcr = np.mean(librosa.feature.zero_crossing_rate(audio))

    return np.hstack([mfcc, chroma, rms, zcr])

df = pd.read_csv(ESC_CSV)

df = df[df["category"].isin([
    "birds", "rain", "wind", "footsteps",
    "siren", "car_horn", "engine", "drilling"
])]

impact_map = {
    "birds": "pleasant",
    "rain": "pleasant",
    "wind": "neutral",
    "footsteps": "neutral",
    "siren": "harmful",
    "car_horn": "harmful",
    "engine": "harmful",
    "drilling": "harmful"
}

df["impact"] = df["category"].map(impact_map)

features, labels = [], []

for _, row in tqdm(df.iterrows(), total=len(df)):
    file_path = os.path.join(AUDIO_PATH, row["filename"])
    try:
        features.append(extract_features(file_path))
        labels.append(row["impact"])
    except:
        pass

X = np.array(features)
y = np.array(labels)

X.shape, y.shape

```
